# Bayesian Reasoning - Medical Diagnosis

```python

disease_probability = 0.10
symptom_probability = 0.80
false_alarm_probability = 0.20

# Bayes Theorem
numerator = disease_probability * symptom_probability

denominator = numerator + (1 - disease_probability) * false_alarm_probability

result = numerator / denominator

print("Bayesian Medical Diagnosis")
print("Probability of Disease:", round(result * 100, 2), "%")

if result > 0.5:
    print("Possible Disease Detected")
else:
    print("Low Probability of Disease")
```

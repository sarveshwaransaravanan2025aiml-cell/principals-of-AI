# Fuzzy Inference - Smart Manufacturing

```python

temperature = 85
vibration = 80

if temperature < 40:
    temp_status = "Low"
elif temperature < 70:
    temp_status = "Medium"
else:
    temp_status = "High"

if vibration < 40:
    vibration_status = "Low"
elif vibration < 70:
    vibration_status = "Medium"
else:
    vibration_status = "High"

print("Smart Manufacturing System")
print("Temperature:", temp_status)
print("Vibration:", vibration_status)

if temp_status == "High" and vibration_status == "High":
    print("Machine Condition: Critical")
    print("Action: Stop Machine")
elif temp_status == "High" or vibration_status == "High":
    print("Machine Condition: Warning")
    print("Action: Check Machine")
else:
    print("Machine Condition: Normal")
    print("Action: Continue Production")
```

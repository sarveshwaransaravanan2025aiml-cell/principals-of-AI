```python
from collections import deque

graph = {
    "START": ["A", "B"],
    "A": ["C"],
    "B": ["D"],
    "C": [],
    "D": ["GOAL"],
    "GOAL": []
}


def bfs(start, goal):

    queue = deque([[start]])
    visited = set()

    while queue:

        path = queue.popleft()
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbour in graph[node]:
                if neighbour not in visited:
                    new_path = path + [neighbour]
                    queue.append(new_path)

    return None


path = bfs("START", "GOAL")

print("BFS Path:")
print(" -> ".join(path))
```

```python
def minimax(scores, is_maximizing):
    if isinstance(scores, int):
        return scores
    if is_maximizing:
        best = -999
        for score in scores:
            value = minimax(score, False)
            best = max(best, value)
        return best
    else:
        best = 999
        for score in scores:
            value = minimax(score, True)
            best = min(best, value)
        return best

game_tree = [[3, 5], [2, 9]]
result = minimax(game_tree, True)
print("Best score:", result)
```

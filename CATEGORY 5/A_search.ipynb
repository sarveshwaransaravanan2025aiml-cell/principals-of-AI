```python
import heapq

def manhattan(state, goal):
    distance = 0
    for tile in range(1, 9):
        current_pos = state.index(tile)
        goal_pos = goal.index(tile)
        current_row, current_col = divmod(current_pos, 3)
        goal_row, goal_col = divmod(goal_pos, 3)
        distance += abs(current_row - goal_row) + abs(current_col - goal_col)
    return distance

def get_neighbors(state):
    neighbors = []
    blank = state.index(0)
    row, col = divmod(blank, 3)
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    for dr, dc in moves:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col
            new_state = list(state)
            new_state[blank], new_state[new_blank] = new_state[new_blank], new_state[blank]
            neighbors.append(tuple(new_state))
    return neighbors

def a_star(start, goal):
    pq = []
    h = manhattan(start, goal)
    heapq.heappush(pq, (h, 0, start))
    g_cost = {start: 0}
    parent = {start: None}
    while pq:
        f, g, current = heapq.heappop(pq)
        print("\nCurrent State:", current)
        print("g(n) =", g)
        print("h(n) =", manhattan(current, goal))
        print("f(n) =", f)
        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            path.reverse()
            return path
        for neighbor in get_neighbors(current):
            new_g = g + 1
            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                h = manhattan(neighbor, goal)
                f = new_g + h
                parent[neighbor] = current
                heapq.heappush(pq, (f, new_g, neighbor))
    return None

start = (1, 2, 3, 0, 4, 6, 7, 5, 8)
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)
path = a_star(start, goal)
print("\n==============================")
print("FINAL PATH")
print("==============================")
for i, state in enumerate(path):
    print("Step", i, ":", state)
print("\nTotal Moves:", len(path) - 1)
```

```python
import random

graph = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D"],
    "D": ["B", "C"]
}

colors = ["Red", "Green", "Blue"]

assignment = {
    "A": "Red",
    "B": "Red",
    "C": "Blue",
    "D": "Blue"
}


def count_conflicts(assignment):
    conflicts = 0

    for region in graph:
        for neighbour in graph[region]:
            if assignment[region] == assignment[neighbour]:
                conflicts += 1

    return conflicts // 2


def local_search():
    for step in range(100):
        conflicts = count_conflicts(assignment)

        print("Step:", step)
        print("Assignment:", assignment)
        print("Conflicts:", conflicts)

        if conflicts == 0:
            print("Solution found!")
            return

        region = random.choice(list(graph.keys()))
        old_color = assignment[region]
        best_color = old_color
        best_conflicts = conflicts

        for color in colors:
            assignment[region] = color
            new_conflicts = count_conflicts(assignment)

            if new_conflicts < best_conflicts:
                best_conflicts = new_conflicts
                best_color = color

        assignment[region] = best_color

    print("Search limit reached.")


local_search()
```

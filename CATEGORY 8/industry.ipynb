```python
facts = ["high_temperature", "high_vibration"]

rules = [
    (["high_temperature"], "temperature_warning"),
    (["high_vibration"], "vibration_warning"),
    (["temperature_warning", "vibration_warning"],
     "maintenance_required"),
    (["maintenance_required"], "stop_machine")
]


def forward_chaining(facts, rules):
    facts = set(facts)
    changed = True

    while changed:
        changed = False

        for conditions, conclusion in rules:
            if all(condition in facts for condition in conditions):
                if conclusion not in facts:
                    facts.add(conclusion)
                    print("New fact:", conclusion)
                    changed = True

    return facts


result = forward_chaining(facts, rules)

print("\nFinal Knowledge Base:")

for fact in sorted(result):
    print(fact)

if "stop_machine" in result:
    print("\nAction: Stop machine and inspect")
elif "maintenance_required" in result:
    print("\nAction: Schedule maintenance")
else:
    print("\nAction: Continue monitoring")
```

```python
facts = ["fever", "cough"]

rules = [
    (["fever", "cough"], "possible_respiratory_infection"),
    (["fever", "body_pain"], "possible_flu_like_illness"),
    (["possible_respiratory_infection", "breathing_difficulty"],
     "urgent_medical_assessment_needed")
]


def forward_chaining(facts, rules):
    facts = set(facts)
    changed = True

    while changed:
        changed = False

        for conditions, conclusion in rules:
            if all(condition in facts for condition in conditions):
                if conclusion not in facts:
                    facts.add(conclusion)
                    print("New fact:", conclusion)
                    changed = True

    return facts


result = forward_chaining(facts, rules)

print("\nFinal facts:")
for fact in sorted(result):
    print(fact)
```

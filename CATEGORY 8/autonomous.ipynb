```python
facts = ["red_signal", "obstacle_ahead"]

rules = [
    (["red_signal"], "stop_vehicle"),
    (["obstacle_ahead"], "stop_vehicle"),
    (["green_signal", "road_clear"], "move_vehicle"),
    (["pedestrian_ahead"], "slow_down")
]


def forward_chaining(facts, rules):
    facts = set(facts)
    changed = True

    while changed:
        changed = False

        for conditions, conclusion in rules:
            if all(condition in facts for condition in conditions):
                if conclusion not in facts:
                    facts.add(conclusion)
                    print("New fact:", conclusion)
                    changed = True

    return facts


result = forward_chaining(facts, rules)

print("\nFinal Knowledge Base:")

for fact in sorted(result):
    print(fact)

if "stop_vehicle" in result:
    print("\nVehicle Action: STOP")
elif "slow_down" in result:
    print("\nVehicle Action: SLOW DOWN")
elif "move_vehicle" in result:
    print("\nVehicle Action: MOVE")
else:
    print("\nVehicle Action: WAIT")
```

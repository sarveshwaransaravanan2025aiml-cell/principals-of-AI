```python
import random
import math

jobs = {
    "A": 3,
    "B": 5,
    "C": 2,
    "D": 4
}


def calculate_cost(schedule):
    total = 0
    completion_time = 0

    for job in schedule:
        completion_time += jobs[job]
        total += completion_time

    return total


def get_neighbor(schedule):
    neighbor = schedule.copy()

    i, j = random.sample(range(len(neighbor)), 2)

    neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

    return neighbor


current = list(jobs.keys())
current_cost = calculate_cost(current)

temperature = 10
cooling_rate = 0.9

print("Initial Schedule:", current)
print("Initial Cost:", current_cost)

for step in range(20):

    neighbor = get_neighbor(current)
    neighbor_cost = calculate_cost(neighbor)

    difference = neighbor_cost - current_cost

    if difference < 0:
        current = neighbor
        current_cost = neighbor_cost
        accepted = True

    else:
        probability = math.exp(-difference / temperature)
        random_value = random.random()

        if random_value < probability:
            current = neighbor
            current_cost = neighbor_cost
            accepted = True
        else:
            accepted = False

    print("\nStep:", step + 1)
    print("Current:", current)
    print("Cost:", current_cost)
    print("Temperature:", round(temperature, 2))
    print("Accepted:", accepted)

    temperature = temperature * cooling_rate

    if temperature < 0.01:
        break


print("\nFINAL SCHEDULE")
print("Schedule:", current)
print("Cost:", current_cost)
```
